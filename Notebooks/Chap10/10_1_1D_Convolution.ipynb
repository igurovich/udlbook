Now let's define the convolutions from figure 10.3.  

```python
def conv1d_custom(signal, kernel, stride=1, dilation=1):
    """
    Performs a 1D convolution operation.

    Args:
        signal (np.array): The input 1D signal.
        kernel (np.array): The 1D convolution kernel (filter).
        stride (int): The step size for moving the kernel across the signal.
        dilation (int): The spacing between kernel elements.

    Returns:
        np.array: The convolved output signal.
    """
    # if not isinstance(signal, np.ndarray) or not isinstance(kernel, np.ndarray):
    #     raise TypeError("Input signal and kernel must be NumPy arrays.")
    # if signal.ndim != 1 or kernel.ndim != 1:
    #     raise ValueError("Input signal and kernel must be 1-dimensional.")

    kernel_size = len(kernel)
    signal_length = len(signal)

    # Calculate the effective kernel size with dilation
    effective_kernel_size = (kernel_size - 1) * dilation + 1

    # Calculate output length
    output_length = (signal_length - effective_kernel_size) // stride + 1
    if output_length <= 0:
        raise ValueError("Output length is not positive. Consider padding or smaller kernel/dilation.")

    output = np.zeros(output_length)

    # Reverse the kernel for convolution (equivalent to cross-correlation with a reversed kernel)
    reversed_kernel = kernel[::-1]

    for i in range(output_length):
        start_index = i * stride
        end_index = start_index + effective_kernel_size

        # Extract the segment of the signal
        signal_segment = signal[start_index:end_index]

        # Apply dilation to the kernel and perform element-wise multiplication
        # This creates a "dilated" version of the kernel for the current segment
        dilated_kernel_applied = np.zeros_like(signal_segment)
        for k_idx, k_val in enumerate(reversed_kernel):
            if k_idx * dilation < len(signal_segment):
                dilated_kernel_applied[k_idx * dilation] = k_val

        output[i] = np.sum(signal_segment * dilated_kernel_applied)

    return output
```
